## Dictionaries

- Recognized by curly brackets {    }

- They are mutable and can be nested like list and tuples

- They have a key and value pair , so they are also called associative pair

- **A given key can only appear once in a dictionary** and both key and value can be any data type - str,int,float,tuple,list

- key and value is seperated by a colon(  :  )

- **Dictionary are unordered so indexing does not work on it**. Instead keys are treated as index. whereas list and tuples are ordered

- when we convert a dictionary to list or a tuple by list() then only key names come in list and tuple not the values

## Code for Even & ODD Number

Save all odd & even numbers till 200 in seprate lists and then print both the list a single time, us for loop with if in one solution and while loop in other.

**Input** -  Iterator with number from 0 to 201

**Logic** -  checking whether number is divisible by 2 or not, if divisible then number is even, else odd. Save the respective number in alloted list.

**Output** - 2 list, one for even number and one for odd number.

There can be multiple solutions to a problem statment, the factor that differentiates the solutions are:
    
    1) Time Complexity: How much time the code takes to execute
    2) Space complexity: How much Memory the code is consuming to give you the output

###  Solution 1

```python
%%time

even_lst=[]
odd_lst=[]

for i in range(0,201):
    if ((i%2)==0):
        even_lst.append(i)
    else:
        odd_lst.append(i)
    
# print('even list',even_lst,'\n\n')
# print('odd list',odd_lst)
```

###  Solution 2

```python
even_lst=[]
odd_lst=[]

for i in range(0,201):
    if ((i%2)==0):
        even_lst.append(i)
    if ((i%2)!=0):
        odd_lst.append(i)
    
print('even list',even_lst,'\n\n')
print('odd list',odd_lst)
```

### USING WHILE LOOP

### Solution 1 

```python
even_lst=[]
odd_lst=[]

i=1

while i<101:
    
    if ((i%2)==0):
        even_lst.append(i)
    else:
        odd_lst.append(i)
        
    i+=1
        
    
print('even list',even_lst,'\n\n')
print('odd list',odd_lst)
```

### Solution 2

```python
even_lst=[]
odd_lst=[]

init=True

i=1

while init:
    
    if ((i%2)==0):
        even_lst.append(i)
        
    else:
        odd_lst.append(i)
    
    i+=1
    
    if i>100:
        init=False
        
print('even list',even_lst,'\n\n')
print('odd list',odd_lst)
```

```python
def product_category(customer_name, product, price):
    if price > 5000:
        category = "Expensive"
    elif 500 <= price <= 1000:
        category = "Affordable"
    else:
        category = "Cheap"
        
    print(f'\n {customer_name} bought {product} and it is {category}')

product_category("Rahul", "Laptop", 55000)
product_category("Priya", "Shoes", 800)
product_category("Aman", "Pen", 50)
```

```python
def most_expensive_product(customers, products, prices):
    max_price = max(prices)
    index1 = prices.index(max_price)

    return {
        "customer": customers[index1],
        "product": products[index1],
        "price": max_price }

customers = ["Rahul", "Priya", "Aman"]
products = ["Laptop", "Mobile", "TV"]
prices = [55000, 25000, 70000]

result = most_expensive_product(customers, products, prices)

print("Customer:", result["customer"])
print("Product:", result["product"])
print("Price:", result["price"])
```

```python
def get_price():
    attempts = 0
    while attempts <= 3:
        try:
            price = float(input("Price: "))
            return price
        except ValueError:
            attempts += 1
            print(f"Invalid price! Attempts left: {3 - attempts}")

    raise ValueError("Maximum attempts exceeded for price input")
```

```python
def most_expensive_product():
    try:
        n = int(input("Enter number of customers: "))
        customers = []
        products = []
        prices = []
        for i in range(n):
            customers.append(input("Customer Name: "))
            products.append(input("Product Name: "))

            p = get_price()
            prices.append(p)

        max_price = max(prices)
        index = prices.index(max_price)
        print("\nMost Expensive Product Details:")
        print("Customer:", customers[index])
        print("Product:", products[index])
        print("Price:", max_price)

    except Exception as e:
        print("\nError:", e)

most_expensive_product()
```

### Nested Function Call

```python
def add(a,b):
    return a+b
    
def sub(a,b):
    return a-b
    
def mul(a,b):
    return a*b
    
def div(a,b):
    return a/b  
```

```python
def collect(c,d):
    print("c :",c,"d :",d)    
    va1=add(c,d)  
    va2=sub(va1,d)
    va3=mul(va2,d)
    va4=div(va3,d)
    
    # return va4
    # return va1,va2,va3,va4
    return locals()
```

```python
output = collect(1,2)
output
```

```python
va4 = output['va4']
print(va4)
```
